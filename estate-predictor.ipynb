# EstatePredictor: Data-Driven House Price Prediction

# Problem Statement

**HomeVista Properties** is a real estate company operating across multiple cities and handling thousands of residential property sales every year.

The company wants to **automate its house pricing process** using Machine Learning.

The objective is to build an intelligent **regression model** that can predict the market price of a house based on its:

* Physical features
* Location
* Condition

## Objective

As a **Machine Learning Engineer**, the task is to:

1. Analyze the historical property dataset.
2. Perform necessary data preprocessing.
3. Train a **Linear Regression** model.
4. Predict house prices based on property characteristics.
5. Evaluate the model's performance.

>The goal is to develop a data-driven system that can help **HomeVista Properties** estimate residential property prices automatically.


```python
# importing libraries

import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import r2_score, mean_absolute_error, mean_squared_error, mean_absolute_percentage_error

print("Libraries imported successfully...Bliss")
```

```python
path = "/kaggle/input/datasets/anushkadabas/house-price-prediction/HousePricePrediction (2).csv"
df = pd.read_csv(path)
print("Dataset loaded successfully")
```

### Data Preprocessing

```python
df = df.drop(['Id'], axis = 1)
```

```python
df.head()
```

```python
df.isnull().sum()
```

```python
df['SalePrice'] = df['SalePrice'].fillna(df['SalePrice'].mean())
```

```python
df.isnull().sum()
```

```python
df = df.dropna()
df.isnull().sum()
```

```python
df.head()
```

```python
cols = ['MSZoning', 'LotConfig', 'BldgType', 'Exterior1st']
df = pd.get_dummies(df, columns=cols, drop_first=True)
```

```python
df
```

### Train Test Split and Train Model

```python
X = df.drop(['SalePrice'], axis=1)
Y = df['SalePrice']

X_train, X_test, y_train, y_test = train_test_split(
    X, Y, test_size=0.2, random_state=0)
```

```python
model = LinearRegression()
model.fit(X_train, y_train)
```

### Model Evalution

```python
y_pred = model.predict(X_test)
```

```python
print("R2 Score:", r2_score(y_test, y_pred))
print("MAE:", mean_absolute_error(y_test, y_pred))
print("RMSE:", np.sqrt(mean_squared_error(y_test, y_pred)))
print(mean_absolute_percentage_error(y_test, y_pred))
```

# Feature Scaling
>to try & improve baseline performance

```python
scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

model.fit(X_train_scaled, y_train)

y_pred = model.predict(X_test_scaled)
```

```python
print("R2 Score:", r2_score(y_test, y_pred))
print("MAE:", mean_absolute_error(y_test, y_pred))
print("RMSE:", np.sqrt(mean_squared_error(y_test, y_pred)))
print(mean_absolute_percentage_error(y_test, y_pred))
```
